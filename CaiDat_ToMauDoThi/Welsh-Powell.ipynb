# Cài đặt thuật toán Welsh-Powell 
## Giải bài toán Tô màu đồ thị



```python
"""
Định dạng input:
- Dòng 1: chứa số đỉnh
- n dòng tiếp theo, mỗi dòng chứa n số (0 hoặc 1), biểu thị ma trận kề của đồ thị
- 1: có cạnh, 0: không có cạnh
"""

import sys

def read_graph(filename):
    try:
        with open(filename, 'r') as f:
       
            n = int(f.readline().strip())
            matrix = []
            for i in range(n):
                row = list(map(int, f.readline().strip().split()))
                if len(row) != n:
                    print(f"Lỗi: Dòng {i+2} có {len(row)} số, cần {n} số để tạo ma trận {n}x{n}")
                    sys.exit(1)
                # Kiểm tra giá trị hợp lệ (0 hoặc 1)
                for val in row:
                    if val not in [0, 1]:
                        print(f"Lỗi: Ma trận kề chứa giá trị không hợp lệ ({val})")
                        sys.exit(1)
                matrix.append(row)
            
            # Chuyển ma trận kề thành danh sách kề
            adj_list = [[] for _ in range(n)]
            for i in range(n):
                for j in range(n):
                    if matrix[i][j] == 1 and i != j:  # Có cạnh và không phải vòng lặp
                        adj_list[i].append(j)
            return adj_list, n
    except FileNotFoundError:
        print(f"Không tìm thấy file {filename}")
        sys.exit(1)

def welsh_powell(adj_list, n):
    """
    Input: Danh sách kề, số đỉnh n
    Output: Danh sách màu của các đỉnh, số màu sử dụng
    """
    # Tính bậc của mỗi đỉnh
    degrees = [(len(adj_list[i]), i) for i in range(n)]
    # Sắp xếp đỉnh theo bậc giảm dần
    degrees.sort(reverse=True)
    # Danh sách màu 
    colors = [-1] * n  # -1 nghĩa là chưa tô màu
    current_color = 0  

    for _, vertex in degrees:
        if colors[vertex] == -1:  # Nếu đỉnh chưa được tô màu
            colors[vertex] = current_color
            # Tô màu cho các đỉnh không kề với vertex, chưa được tô màu
            for other_vertex in range(n):
                if colors[other_vertex] == -1:  # Chưa tô màu
                    # Kiểm tra xem other_vertex có kề với vertex không
                    can_color = True
                    for neighbor in adj_list[other_vertex]:
                        if colors[neighbor] == current_color:
                            can_color = False
                            break
                    if can_color:
                        colors[other_vertex] = current_color
            current_color += 1  # Chuyển sang màu tiếp theo

    return colors, current_color

def main():
   
    filename = "input.txt"
    adj_list, n = read_graph(filename)

    colors, num_colors = welsh_powell(adj_list, n)

    print("Kết quả tô màu đồ thị (Welsh-Powell):")
    for i in range(n):
        print(f"Đỉnh {i}: Màu {colors[i]}")
    print(f"Số màu sử dụng: {num_colors}")

main()
```
