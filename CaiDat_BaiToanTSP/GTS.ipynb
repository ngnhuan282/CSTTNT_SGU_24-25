# Cài đặt thuật toán GTS1, GTS2


```python
"""
Cài đặt thuật toán GTS1 và GTS2
"""

import sys

def read_matrix(filename):
    """
    Đọc ma trận chi phí từ file
    Input: Tên file (input.txt)
    Output: Ma trận chi phí và số thành phố n
    """
    matrix = []
    try:
        with open(filename, 'r') as f:
            n = int(f.readline().strip())  # Số thành phố
            for _ in range(n):
                row = list(map(int, f.readline().strip().split()))
                matrix.append(row)
        return matrix, n
    except FileNotFoundError:
        print(f"Không tìm thấy file {filename}")
        sys.exit(1)

def GTS1(matrix, n, start):
    """
    Thuật toán GTS1: Tìm hành trình TSP từ một điểm xuất phát
    Input: Ma trận chi phí, số thành phố n, điểm xuất phát start
    Output: Hành trình (tour) và chi phí (cost)
    """
    visited = [False] * n  # Mảng đánh dấu thành phố đã thăm
    tour = [start]  # Lưu hành trình
    visited[start] = True
    cost = 0  # Tổng chi phí
    current = start
    count = 1  # Số thành phố đã thăm

    while count < n:
        min_cost = float('inf')
        next_city = -1
        # Tìm thành phố chưa thăm có chi phí nhỏ nhất từ current
        for j in range(n):
            if not visited[j] and matrix[current][j] != 0 and matrix[current][j] < min_cost:
                min_cost = matrix[current][j]
                next_city = j
        if next_city == -1:  # Không tìm được thành phố tiếp theo
            return None, float('inf')  # Không có lời giải
        cost += min_cost
        tour.append(next_city)
        visited[next_city] = True
        current = next_city
        count += 1

    # Kiểm tra xem có thể quay lại điểm xuất phát
    if matrix[current][start] != 0:
        cost += matrix[current][start]
        tour.append(start)
        return tour, cost
    return None, float('inf')  # Không có đường về

def GTS2(matrix, n):
    """
    GTS2: Thử GTS1 với mọi điểm xuất phát và chọn hành trình tốt nhất
    Input: Ma trận chi phí, số thành phố n
    Output: Hành trình tốt nhất (tour) và chi phí nhỏ nhất (cost)
    """
    best_tour = None
    best_cost = float('inf')

    for start in range(n):
        tour, cost = GTS1(matrix, n, start)
        if tour and cost < best_cost:
            best_cost = cost
            best_tour = tour

    return best_tour, best_cost

def main():
    # Đọc ma trận từ file input.txt
    filename = "input.txt"
    matrix, n = read_matrix(filename)

    # Chạy GTS1 với điểm xuất phát là thành phố 0
    print("1. GTS1 (xuất phát từ thành phố 0):")
    tour, cost = GTS1(matrix, n, 0)
    if tour:
        print(f"Hành trình: {' -> '.join(map(str, tour))}")
        print(f"Chi phí: {cost}")
    else:
        print("Không tìm được hành trình khả thi")

    # Chạy GTS2
    print("\n2. GTS2 (tìm hành trình tốt nhất):")
    tour, cost = GTS2(matrix, n)
    if tour:
        print(f"Hành trình tốt nhất: {' -> '.join(map(str, tour))}")
        print(f"Chi phí nhỏ nhất: {cost}")
    else:
        print("Không tìm được hành trình khả thi")

main()
```
